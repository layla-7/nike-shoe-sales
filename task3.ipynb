```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("/content/Walmart_Sales.csv")
df.head(3)
```

```python
plt.figure(figsize=(10,6))
plt.hist(df["Temperature"], bins=20, color="Green", alpha=0.7, edgecolor="Black")
plt.title("Temperature & Frequency")
plt.xlabel("Temperature")
plt.ylabel("Frequency")
plt.grid(axis="y")
xmin, xmax = plt.xlim()
print("The chart currently goes from", xmin, "to", xmax)
ymin, ymax = plt.ylim()
print("The chart currently goes from", ymin, "to", ymax)
plt.savefig("Temperature & Frequency Diagram.png")
plt.show()
```

```python
store_sales = df.groupby("Store")["Weekly_Sales"].sum().sort_values(ascending=False)
top5 = store_sales.head(5)
top5["Others"] = store_sales.iloc[5:].sum()
top5.index = [f"Store {i}" if i != "Others" else "Others" for i in top5.index]

plt.figure(figsize=(10,6))
plt.pie(top5, labels=top5.index, autopct='%1.1f%%', startangle=90)
plt.title("Weekly Sales by Store")
plt.axis("equal")
plt.savefig("Weekly Sales by Store.png")
plt.show()
```
