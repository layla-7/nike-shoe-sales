## Load Nike dataset

```python
import pandas as pd
from pathlib import Path

path = Path('nike_shoes_sales.csv')
if not path.exists():
    from google.colab import files
    print('Upload nike_shoes_sales.csv')
    uploaded = files.upload()
    path = Path('nike_shoes_sales.csv')
    if not path.exists():
        raise FileNotFoundError('Please upload nike_shoes_sales.csv')
df = pd.read_csv(path)
print('Dataset loaded:', df.shape[0], 'rows and', df.shape[1], 'columns')
```

## 1. head()
Preview first five records.

```python
display(df.head())
```

## 2. tail()
Preview last five records.

```python
display(df.tail())
```

## 3. info()
Show data types and non-null counts.

```python
df.info()
```

## 4. describe()
Summarize numeric columns.

```python
display(df.describe())
```

## 5. isnull()
Identify missing values.

```python
display(df.isnull().sum())
```

## 6. fillna()
Fill missing descriptions in a working copy.

```python
clean = df.copy()
clean["description"] = clean["description"].fillna("Not provided")
display(clean[["product_name", "description"]].head())
```

## 7. drop_duplicates()
Remove duplicate rows in a working copy.

```python
unique = df.drop_duplicates()
print("Rows after removing duplicates:", len(unique))
```

## 8. value_counts()
Count products by brand.

```python
display(df["brand"].value_counts())
```

## 9. sort_values()
Find products with the highest ratings.

```python
display(df.sort_values("rating", ascending=False)[["product_name", "rating"]].head())
```

## 10. groupby()
Summarize average sale price by brand.

```python
display(df.groupby("brand")["sale_price"].mean())
```

## 11. mean()
Calculate mean sale price.

```python
print(df["sale_price"].mean())
```

## 12. median()
Calculate median sale price.

```python
print(df["sale_price"].median())
```

## 13. max()
Find maximum sale price.

```python
print(df["sale_price"].max())
```

## 14. min()
Find minimum sale price.

```python
print(df["sale_price"].min())
```

## 15. nunique()
Count distinct products.

```python
print(df["product_id"].nunique())
```
