## Load Nike dataset

```python
import pandas as pd
from pathlib import Path

path = Path('nike_shoes_sales.csv')
if not path.exists():
    from google.colab import files
    print('Upload nike_shoes_sales.csv')
    uploaded = files.upload()
    path = Path('nike_shoes_sales.csv')
    if not path.exists():
        raise FileNotFoundError('Please upload nike_shoes_sales.csv')
df = pd.read_csv(path)
print('Dataset loaded:', df.shape[0], 'rows and', df.shape[1], 'columns')
```

## Matplotlib: 10 functions and 2 charts
Functions used: `figure()`, `hist()`, `title()`, `xlabel()`, `ylabel()`, `grid()`, `tight_layout()`, `show()`, `bar()`, and `xticks()`.

## Chart 1: Distribution of Nike shoe sale prices

```python
import matplotlib.pyplot as plt
prices = df['sale_price'].dropna()
plt.figure(figsize=(9, 5))  # 1. figure()
plt.hist(prices, bins=15)  # 2. hist()
plt.title('Nike Shoe Sale Price Distribution')  # 3. title()
plt.xlabel('Sale price (dataset units)')  # 4. xlabel()
plt.ylabel('Number of listings')  # 5. ylabel()
plt.grid(axis='y', alpha=0.3)  # 6. grid()
plt.tight_layout()  # 7. tight_layout()
plt.show()  # 8. show()
```

## Chart 2: Number of Nike shoe listings by rating
The chart counts product listings, not individual customer purchases.

```python
ratings = df['rating'].dropna().round(1).value_counts().sort_index()
plt.figure(figsize=(10, 5))
plt.bar([str(x) for x in ratings.index], ratings.values)  # 9. bar()
plt.title('Nike Shoe Listings by Customer Rating')
plt.xlabel('Rating')
plt.ylabel('Number of listings')
plt.xticks(rotation=45)  # 10. xticks()
plt.tight_layout()
plt.show()
```
